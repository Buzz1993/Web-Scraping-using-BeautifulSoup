```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
quote_data = []

for pages in range(1,11):
    website = f"https://quotes.toscrape.com/page/{pages}/"
    
    r = requests.get(website)
    soup = BeautifulSoup(r.content, "html.parser")
    quote = soup.find_all("div", class_ = "quote")
    
    quote_text_t = None
    author_name_t = None
    tags_t = None
    author_birth_date_t = None
    born_place_t = None
    about_author_t = None

    
    for item in quote:
        try:
            quote_text_t = item.find('span').get_text(strip=True)
        except AttributeError:
            pass

        try:
            author_name_t = item.find('small').get_text()
        except AttributeError:
            pass
            
        try:
            tags_t = item.find('meta')['content']
        except AttributeError:
            pass
            
        half_link = item.find('a')['href']
        full_link = (f"https://quotes.toscrape.com/{half_link}")
        inner_website = full_link
        r_inner = requests.get(inner_website)
        soup_inner = BeautifulSoup(r_inner.content, "html.parser")
        inner_author_name = soup_inner.find('h3').get_text()
        if inner_author_name == author_name_t:
            try:
                author_birth_date_t = soup_inner.find('span').get_text()
            except AttributeError:
                None

            try:
                born_place_t = soup_inner.find('span', class_ = "author-born-location").get_text()
            except AttributeError:
                None

            try:
                about_author_t  = soup_inner.find('div', class_ = 'author-description').get_text()
            except AttributeError:
                pass
    
        all_features = {
            "quote": quote_text_t,
            "author_name": author_name_t,
            "tags": tags_t,
            "author_birth_date": author_birth_date_t,
            "born_place": born_place_t,
            "about_author": about_author_t
        }
    
        quote_data.append(all_features)
    
df = pd.DataFrame(quote_data)
df.to_csv("Quote_data.csv", index=False)
```
